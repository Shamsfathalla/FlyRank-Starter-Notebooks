Decision:

My research question is: Which content pages should be reviewed first for a potential refresh or other content action?
The decision is how a content/SEO team should prioritize its limited review capacity.

Action:

A human reviewer would use the ranked list to investigate the highest-priority pages and decide whether to refresh, expand, protect, monitor, or take another appropriate action.

Cost of a Wrong Call:
